```python
import numpy as numpy
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score,mean_squared_error
df=pd.read_csv("BMW_Cars.csv")

df=df.drop("Unnamed: 0",axis=1)
df["Auction Rating"]=df["Auction Rating"].fillna(df["Auction Rating"].mean(),inplace=True)
df = df.dropna(subset=["Price (PKR)"])
df.isnull().sum()

df=pd.get_dummies(df,columns=["Car Name","Fuel Type","Transmission","Engine Unit"],drop_first=True)
df.head()
x=df.drop("Price (PKR)",axis=1)
y=df["Price (PKR)"]
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(X_train,Y_train)
y_pred=model.predict(X_test)
mae=mean_absolute_error(Y_test,y_pred)
rmse = mean_squared_error(Y_test, y_pred) ** 0.5
r2 = r2_score(Y_test, y_pred)
print("Mean Absolute Error:", mae)
print("Root Mean Squared Error:", rmse)
print("R2 Score:", r2)


```
